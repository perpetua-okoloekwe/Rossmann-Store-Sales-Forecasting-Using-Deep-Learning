## Missing Value Treatment Strategy

### Promo2 Variables

- Promo2SinceWeek
- Promo2SinceYear
- PromoInterval

occur for stores where Promo2 = 0.

### Competition Variables

Missing values in:

- CompetitionOpenSinceMonth
- CompetitionOpenSinceYear

appear to represent unavailable competition start dates rather than the absence of competition, since CompetitionDistance remains available for many affected stores.

Appropriate imputation strategies will be applied during preprocessing while preserving business meaning.

## Missing Value Treatment

Missing values were handled using business-informed strategies rather than applying a single imputation method across all variables.

### CompetitionDistance

Only three records contained missing competition distance values. These values were imputed using the median distance because the variable contains extreme values and the median is more robust to outliers.

### Competition Opening Dates

Missing values in CompetitionOpenSinceMonth and CompetitionOpenSinceYear were treated as unknown competition start dates rather than missing competitors.

Sentinel values were used to preserve this distinction.

### Promo2 Variables

Investigation showed that missing values in Promo2SinceWeek, Promo2SinceYear, and PromoInterval occur for stores where Promo2 = 0.

These were treated as structural missing values and encoded using sentinel values and categorical indicators to preserve business meaning.

```python
# CompetitionDistance
df["CompetitionDistance"] = df[
    "CompetitionDistance"
].fillna(
    df["CompetitionDistance"].median()
)

# Competition dates
df["CompetitionOpenSinceMonth"] = df[
    "CompetitionOpenSinceMonth"
].fillna(0)

df["CompetitionOpenSinceYear"] = df[
    "CompetitionOpenSinceYear"
].fillna(0)

# Promo2 dates
df["Promo2SinceWeek"] = df[
    "Promo2SinceWeek"
].fillna(0)

df["Promo2SinceYear"] = df[
    "Promo2SinceYear"
].fillna(0)

# Promo interval
df["PromoInterval"] = df[
    "PromoInterval"
].fillna("None")
```

## Temporal Feature Engineering

Time is the most important source of information in forecasting problems.

The original Date variable contains valuable temporal information that may not be directly accessible to machine learning and deep learning models.

To capture seasonal, monthly, quarterly, and yearly patterns, additional time-based features will be extracted from the Date variable.

These features will help the models learn recurring temporal behaviors that influence sales performance.


```python
# Extract temporal features

df["Year"] = df["Date"].dt.year

df["Month"] = df["Date"].dt.month

df["Day"] = df["Date"].dt.day

df["WeekOfYear"] = df["Date"].dt.isocalendar().week.astype(int)

df["Quarter"] = df["Date"].dt.quarter

df["DayOfYear"] = df["Date"].dt.dayofyear

df["IsMonthStart"] = df["Date"].dt.is_month_start.astype(int)

df["IsMonthEnd"] = df["Date"].dt.is_month_end.astype(int)
```

## Categorical Encoding

Several variables contain categorical information that cannot be used directly by machine learning and deep learning models.

The following variables were identified as categorical:

- StoreType
- Assortment
- StateHoliday
- PromoInterval

Label Encoding was selected because it provides a compact numerical representation, simplifies deployment, and allows the same encoding mappings to be reused in future Streamlit applications.

The encoding mappings will be preserved to ensure consistency between model training and future inference workflows.

```python
from sklearn.preprocessing import LabelEncoder

# Create encoders
storetype_encoder = LabelEncoder()
assortment_encoder = LabelEncoder()
stateholiday_encoder = LabelEncoder()
promointerval_encoder = LabelEncoder()
```

```python
df["StoreType"] = storetype_encoder.fit_transform(
    df["StoreType"]
)

df["Assortment"] = assortment_encoder.fit_transform(
    df["Assortment"]
)

df["StateHoliday"] = stateholiday_encoder.fit_transform(
    df["StateHoliday"]
)

df["PromoInterval"] = promointerval_encoder.fit_transform(
    df["PromoInterval"]
)
```

```python
print(
    dict(
        zip(
            storetype_encoder.classes_,
            storetype_encoder.transform(
                storetype_encoder.classes_
            )
        )
    )
)

print(
    dict(
        zip(
            assortment_encoder.classes_,
            assortment_encoder.transform(
                assortment_encoder.classes_
            )
        )
    )
)

print(
    dict(
        zip(
            stateholiday_encoder.classes_,
            stateholiday_encoder.transform(
                stateholiday_encoder.classes_
            )
        )
    )
)

print(
    dict(
        zip(
            promointerval_encoder.classes_,
            promointerval_encoder.transform(
                promointerval_encoder.classes_
            )
        )
    )
)
```

### Sequence Design

Sequences will be generated separately for each store to preserve store-specific sales histories.

Each sequence will contain 30 consecutive days of historical observations, while the target will be the sales value on the following day.

A sliding-window approach will be used to generate multiple training examples from each store's historical data.

```python
def create_sequences(X, y, lookback=30):

    X_seq = []
    y_seq = []

    for i in range(len(X) - lookback):

        X_seq.append(
            X[i:i + lookback]
        )


        y_seq.append(
            y[i + lookback]
        )

    return (
        np.array(X_seq),
        np.array(y_seq)
    )
```

```python
X_train_sequences = []
y_train_sequences = []

for store_id in train_df["Store"].unique():

    store_mask = train_df["Store"] == store_id

    X_store = X_train_scaled[store_mask]

    y_store = y_train_scaled[store_mask]

    if len(X_store) > 30:

        X_seq, y_seq = create_sequences(
            X_store,
            y_store,
            lookback=30
        )

        X_train_sequences.append(X_seq)
        y_train_sequences.append(y_seq)
```

```python
X_train_seq = np.concatenate(
    X_train_sequences,
    axis=0
)

y_train_seq = np.concatenate(
    y_train_sequences,
    axis=0
)

print(X_train_seq.shape)
print(y_train_seq.shape)
```

```python
import torch
import torch.nn as nn

class RNNModel(nn.Module):

    def __init__(
        self,
        input_size,
        hidden_size,
        num_layers,
        output_size
    ):
        super().__init__()

        self.rnn = nn.RNN(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True
        )

        self.fc = nn.Linear(
            hidden_size,
            output_size
        )

    def forward(self, x):

        out, _ = self.rnn(x)

        out = self.fc(
            out[:, -1, :]
        )

        return out


input_size = 22
hidden_size = 64
num_layers = 2
output_size = 1

model = RNNModel(
    input_size,
    hidden_size,
    num_layers,
    output_size
)
```

## Long Short-Term Memory (LSTM)

LSTMs were developed to address the limitations of traditional RNNs.

While RNNs can struggle to retain information across longer sequences, LSTMs use memory cells and gating mechanisms to selectively retain or forget information.

This allows the model to learn long-term dependencies within sequential data.

In this project, the LSTM will learn from the previous 30 days of sales-related information to predict the next day's sales.

```python
class LSTMModel(nn.Module):

    def __init__(
        self,
        input_size,
        hidden_size,
        num_layers,
        output_size
    ):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True
        )

        self.fc = nn.Linear(
            hidden_size,
            output_size
        )

    def forward(self, x):

        out, _ = self.lstm(x)

        out = self.fc(
            out[:, -1, :]
        )

        return out
```

## Attention-Based Model

Attention mechanisms allow a model to dynamically focus on the most relevant parts of a sequence when making predictions.

Rather than treating all historical observations equally, attention assigns different importance weights to different time steps.

This enables the model to learn which historical periods contribute most strongly to future sales predictions.

```python
class AttentionModel(nn.Module):

    def __init__(
        self,
        input_size,
        hidden_size,
        output_size
    ):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size,
            hidden_size,
            batch_first=True
        )

        self.attention = nn.Linear(
            hidden_size,
            1
        )

        self.fc = nn.Linear(
            hidden_size,
            output_size
        )

    def forward(self, x):

        lstm_out, _ = self.lstm(x)

        attention_weights = torch.softmax(
            self.attention(lstm_out),
            dim=1
        )

        context_vector = torch.sum(
            attention_weights * lstm_out,
            dim=1
        )

        output = self.fc(
            context_vector
        )

        return output
```

## Transformer Model

Transformers extend the attention mechanism by relying entirely on attention-based computations rather than recurrent processing.

Unlike RNNs and LSTMs, Transformers process all time steps in parallel and learn relationships between observations using self-attention mechanisms.

This architecture enables efficient learning of complex temporal dependencies across long sequences.

```python
class TransformerModel(nn.Module):

    def __init__(
        self,
        input_size,
        num_heads,
        hidden_dim,
        num_layers,
        output_size
    ):
        super().__init__()

        self.input_projection = nn.Linear(
            input_size,
            hidden_dim
        )

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=hidden_dim,
            nhead=num_heads,
            batch_first=True
        )

        self.transformer = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers
        )

        self.fc = nn.Linear(
            hidden_dim,
            output_size
        )

    def forward(self, x):

        x = self.input_projection(x)

        x = self.transformer(x)

        x = x[:, -1, :]

        return self.fc(x)
```

```python
transformer_model = TransformerModel(
    input_size=22,
    num_heads=4,
    hidden_dim=64,
    num_layers=2,
    output_size=1
)

transformer_model = transformer_model.to(device)

print(transformer_model)
```

| RNN | **0.0602** | **0.0783** |

The baseline model produced the highest forecasting error, indicating that simple persistence-based forecasting was insufficient for capturing the underlying sales patterns present in the dataset.

All sequence learning models achieved substantial improvements over the baseline, confirming that historical sales information contains meaningful temporal patterns that can be leveraged for forecasting.

Among the evaluated architectures, the RNN achieved the lowest MAE and RMSE values. Under the current experimental setup, the RNN demonstrated the strongest ability to learn useful sequential patterns from the previous 30 days of sales history.

The Attention model and Transformer model also performed well and significantly outperformed the baseline. However, neither surpassed the RNN. The Attention model provided slightly better performance than the LSTM, suggesting that assigning varying importance to different time steps may offer advantages over relying solely on recurrent memory mechanisms.

Although LSTMs, Attention Networks, and Transformers are generally considered more advanced architectures, their advantages may not have been fully realized in this experiment due to the use of a reduced training subset, limited training epochs, and minimal hyperparameter tuning.

Therefore, the findings should be interpreted as results obtained under the specific experimental conditions rather than a definitive ranking of forecasting architectures.

# Recommendations

Based on the results of this study, the following recommendations are proposed:

1. Use Sequence-Based Models for Forecasting

The substantial performance improvement achieved by all deep learning models compared to the baseline demonstrates the value of sequence learning approaches for retail sales forecasting.

2. Consider RNNs as Strong Baseline Deep Learning Models

Despite being the simplest architecture evaluated, the RNN produced the best forecasting performance under the current experimental setup. RNNs should therefore be considered as strong baseline sequence models before moving to more complex architectures.

3. Perform Further Hyperparameter Optimization

Additional experiments involving learning rate tuning, hidden layer sizes, batch sizes, and sequence lengths may further improve model performance.

4. Train Models Using Larger Datasets

The comparison was performed using a reduced subset of 10,000 training sequences due to computational constraints. Future studies should evaluate the architectures using the complete dataset to obtain more representative results.

5. Increase Training Duration

Training all models for only five epochs may have limited their ability to learn complex temporal relationships. Additional training epochs combined with early stopping techniques may improve performance.

6. Explore Advanced Forecasting Features

Future versions of the project could incorporate:

- Rolling averages
- Lag features
- Seasonal indicators
- Store-level trend information
- External economic variables

7. Deploy the Best Model for Real-Time Forecasting

The trained RNN model can serve as the foundation for a future Streamlit application that allows users to generate sales forecasts from historical store data in an interactive environment.

# Constraints Encountered

Several constraints were encountered during the development and evaluation of the forecasting models.

## Computational Resource Constraints

The original sequence dataset contained:

- 779,714 training sequences
- 170,595 testing sequences

Training deep learning architectures on the full dataset exceeded the available memory resources of the Google Colab environment, resulting in multiple session crashes during model training.

To ensure successful model execution, a reduced training subset of 10,000 sequences was used for model development and comparison.

## Limited Training Duration

Due to computational limitations, all deep learning models were trained for only five epochs.

While this enabled model comparison within the available resources, longer training durations may have allowed the models to learn more complex temporal patterns and potentially achieve improved forecasting performance.

## Limited Hyperparameter Tuning

The models were trained using a common set of hyperparameters to maintain consistency across experiments.

No extensive hyperparameter optimization was performed for:

- Hidden layer dimensions
- Learning rates
- Batch sizes
- Number of layers
- Sequence lengths

As a result, the reported performance may not represent the maximum achievable performance of each architecture.

## Reduced Dataset Usage

The comparison was conducted using a subset of the available training sequences.

More advanced models such as LSTMs, Attention Networks, and Transformers often benefit from substantially larger training datasets. Consequently, the reduced sample size may have limited the ability of these architectures to fully demonstrate their strengths.

## Experimental Rather Than Production Configuration

The primary objective of this project was educational and exploratory rather than production deployment.

The focus was placed on understanding:

- Time-series forecasting workflows
- Sequence generation
- RNN architectures
- LSTM architectures
- Attention mechanisms
- Transformer architectures

Therefore, the project emphasized learning and model comparison rather than maximizing predictive performance.
